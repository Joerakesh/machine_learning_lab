## Aim
Perform K-Means clustering on the heart disease dataset and use the Elbow Method to choose a suitable number of clusters.

## Dataset
Expected path: `heart_disease_dataset.csv`. Obtain the dataset from an authorised source and place it at the expected path. See the [dataset guide](../datasets/README.md).

## Algorithm / Workflow
1. Load data and remove the Heart Disease target.
2. Scale numeric features and one-hot encode categorical features.
3. Run K-Means for K=1 through 10 and record inertia.
4. Plot the Elbow curve.
5. Fit the final model with selected K.
6. Add cluster labels and count records per cluster.

## Implementation
The following code is copied from the existing `ex-13.py` program and split into notebook cells for interactive execution. The original Python program is unchanged.

Run cells from top to bottom. Outputs and charts are generated when executed with the dataset and required libraries available.

```python
# EXERCISE – 13 (K-MEANS CLUSTERING)

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.cluster import KMeans

df = pd.read_csv("https://raw.githubusercontent.com/Joerakesh/machine_learning_lab/refs/heads/main/datasets/heart_disease_dataset.csv")

print("Dataset Shape:", df.shape)
print("\nFirst 5 records:")
print(df.head())

X = df.drop("Heart Disease", axis=1)

numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_features = X.select_dtypes(
    include=["object"]
).columns

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

X_processed = preprocessor.fit_transform(X)

# Check whether X_processed is sparse or already a NumPy array
if hasattr(X_processed, "toarray"):
    X_processed = X_processed.toarray()

print("\nProcessed Data Shape:")
print(X_processed.shape)

# ------------------------------------------------------------
# Step 7: Apply Elbow Method
# ------------------------------------------------------------

inertia = []

K = range(1, 11)

for k in K:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_processed)

    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))

plt.plot(
    K,
    inertia,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal K")

plt.xticks(K)

plt.grid(True)

plt.show()

# Change this value after observing the elbow plot.
optimal_k = 3

print("\nSelected Optimal K:", optimal_k)

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_processed)

df["Cluster"] = clusters

print("\nClustered Dataset:")
print(df.head(10))

print("\nNumber of records in each cluster:")
print(df["Cluster"].value_counts().sort_index())
```
