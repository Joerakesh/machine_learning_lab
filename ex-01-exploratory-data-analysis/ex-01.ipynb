```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = "https://raw.githubusercontent.com/Joerakesh/machine_learning_lab/refs/heads/main/datasets/titanic.csv"
```

```python
df = pd.read_csv(data)
```

```python
df.head()
```

```python
df.tail()
```

```python
df.sample(5)
```

```python
print("Rows and Columns:", df.shape)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.size
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100
```

```python
print(missing_percentage.sort_values(ascending=False))
```

```python
plt.figure(figsize=(10, 6))
```

```python
sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False,
    cmap="viridis"
)
```

```python
plt.title("Missing Values Heatmap")
plt.show()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df["Cabin"] = df["Cabin"].fillna("Unknown")
```

```python
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
print("Shape after removing duplicates:", df.shape)
```

```python
plt.figure(figsize=(6, 4))
```

```python
sns.countplot(data=df, x="Survived")
```

```python
plt.title("Passenger Survival Count")
plt.xlabel("Survived")
plt.ylabel("Number of Passengers")
```

```python
plt.show()
```

```python
survival_counts = df["Survived"].value_counts()
```

```python
plt.figure(figsize=(6, 6))
```

```python
plt.pie(
    survival_counts,
    labels=["Did Not Survive", "Survived"],
    autopct="%1.1f%%",
    startangle=90
)
```

```python
plt.title("Survival Distribution")
```

```python
plt.show()
```

```python
plt.figure(figsize=(6, 4))
```

```python
sns.countplot(data=df, x="Pclass")
```

```python
plt.title("Passenger Class Distribution")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
```

```python
plt.show()
```

```python
plt.figure(figsize=(6, 4))
```

```python
sns.countplot(data=df, x="Sex")
```

```python
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
```

```python
plt.show()
```

```python
plt.figure(figsize=(7, 5))
```

```python
sns.countplot(
    data=df,
    x="Sex",
    hue="Survived"
)
```

```python
plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
```

```python
plt.show()
```

```python
plt.figure(figsize=(7, 5))
```

```python
sns.countplot(
    data=df,
    x="Pclass",
    hue="Survived"
)
```

```python
plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
```

```python
plt.show()
```

```python
plt.figure(figsize=(8, 5))
```

```python
sns.boxplot(
    data=df,
    x="Survived",
    y="Fare"
)
```

```python
plt.title("Fare Distribution by Survival")
plt.xlabel("Survived")
plt.ylabel("Fare")
```

```python
plt.show()
```

```python
g = sns.catplot(
    data=df,
    x="Pclass",
    hue="Survived",
    col="Sex",
    kind="count",
    height=5,
    aspect=1
)
```

```python
g.fig.suptitle(
    "Survival by Passenger Class and Gender",
    y=1.05
)
```

```python
plt.show()
```

```python
plt.figure(figsize=(9, 6))
```

```python
sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived",
    style="Sex",
    size="Pclass",
    sizes=(50, 200)
)
```

```python
plt.title("Age vs Fare by Survival, Gender and Class")
plt.xlabel("Age")
plt.ylabel("Fare")
```

```python
plt.show()
```

```python
numeric_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare"
]
```

```python
correlation = df[numeric_columns].corr()
```

```python
print(correlation)
```

```python
plt.figure(figsize=(9, 7))
```

```python
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
```

```python
plt.title("Correlation Heatmap")
```

```python
plt.show()
```
