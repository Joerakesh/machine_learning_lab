```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    roc_auc_score
)
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/Joerakesh/machine_learning_lab/refs/heads/main/datasets/heart_disease_dataset.csv")
```

```python
print("Dataset Shape:", df.shape)
```

```python
print("\nFirst 5 Records:")
print(df.head())
```

```python
print("\nDataset Information:")
print(df.info())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nTarget Distribution:")
print(df["Heart Disease"].value_counts())
```

```python
X = df.drop("Heart Disease", axis=1)
```

```python
y = df["Heart Disease"]
```

```python
numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns
```

```python
categorical_features = X.select_dtypes(
    include=["object"]
).columns
```

```python
print("Numerical Features:")
print(list(numerical_features))
```

```python
print("\nCategorical Features:")
print(list(categorical_features))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training Data:", X_train.shape)
print("Testing Data :", X_test.shape)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
```

```python
K = 5
knn = KNeighborsClassifier(
    n_neighbors=5
)
```

```python
knn.fit(
    X_train_processed,
    y_train
)
```

```python
y_pred = knn.predict(
    X_test_processed
)
```

```python
print("Predicted Values:")
print(y_pred[:20])
```

```python
print(
    classification_report(
        y_test,
        y_pred
    )
)
```

```python
k_values = [1, 3, 5, 7, 9, 11, 15, 21]
```

```python
accuracy_values = []
precision_values = []
recall_values = []
f1_values = []
```

```python
for k in k_values:
```

```python
knn = KNeighborsClassifier(
        n_neighbors=k
    )
```

```python
knn.fit(
        X_train_processed,
        y_train
    )
```

```python
y_pred = knn.predict(
        X_test_processed
    )
```

```python
accuracy_values.append(
        accuracy_score(
            y_test,
            y_pred
        )
    )
```

```python
precision_values.append(
        precision_score(
            y_test,
            y_pred
        )
    )
```

```python
recall_values.append(
        recall_score(
            y_test,
            y_pred
        )
    )
```

```python
f1_values.append(
        f1_score(
            y_test,
            y_pred
        )
    )
```

```python
results = pd.DataFrame({
    "K": k_values,
    "Accuracy": accuracy_values,
    "Precision": precision_values,
    "Recall": recall_values,
    "F1 Score": f1_values
})
```

```python
print(results)
```

```python
plt.figure(figsize=(8, 5))
```

```python
plt.plot(
    k_values,
    accuracy_values,
    marker="o"
)
```

```python
plt.xlabel("K Value")
plt.ylabel("Accuracy")
```

```python
plt.title("Effect of K Value on KNN Accuracy")
```

```python
plt.xticks(k_values)
```

```python
plt.grid()
```

```python
plt.show()
```

```python
best_index = np.argmax(f1_values)
```

```python
best_k = k_values[best_index]
```

```python
print("Best K:", best_k)
```

```python
print(
    "Best Accuracy:",
    accuracy_values[best_index]
)
```

```python
print(
    "Best Precision:",
    precision_values[best_index]
)
```

```python
print(
    "Best Recall:",
    recall_values[best_index]
)
```

```python
print(
    "Best F1 Score:",
    f1_values[best_index]
)
```

```python
final_knn = KNeighborsClassifier(
    n_neighbors=best_k
)
```

```python
final_knn.fit(
    X_train_processed,
    y_train
)
```

```python
final_pred = final_knn.predict(
    X_test_processed
)
```

```python
final_accuracy = accuracy_score(
    y_test,
    final_pred
)
```

```python
final_precision = precision_score(
    y_test,
    final_pred
)
```

```python
final_recall = recall_score(
    y_test,
    final_pred
)
```

```python
final_f1 = f1_score(
    y_test,
    final_pred
)
```

```python
final_cm = confusion_matrix(
    y_test,
    final_pred
)
```

```python
TN, FP, FN, TP = final_cm.ravel()
```

```python
final_specificity = TN / (TN + FP)
```

```python
final_probability = final_knn.predict_proba(
    X_test_processed
)[:, 1]
```

```python
final_auc = roc_auc_score(
    y_test,
    final_probability
)
```

```python
print("\n======================================")
print("FINAL KNN MODEL")
print("======================================")
```

```python
print("Best K       :", best_k)
print("Accuracy     :", final_accuracy)
print("Precision    :", final_precision)
print("Recall       :", final_recall)
print("F1 Score     :", final_f1)
print("Specificity  :", final_specificity)
print("AUC          :", final_auc)
```
