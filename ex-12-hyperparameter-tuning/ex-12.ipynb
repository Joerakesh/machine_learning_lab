```python
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/Joerakesh/machine_learning_lab/refs/heads/main/datasets/heart_disease_dataset.csv")
```

```python
print("Dataset Shape:", df.shape)
print(df.head())
```

```python
X = df.drop("Heart Disease", axis=1)
y = df["Heart Disease"]
```

```python
numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns
```

```python
categorical_features = X.select_dtypes(
    include=["object"]
).columns
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)
```

```python
pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("svm", SVC())
    ]
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
param_grid = {
    "svm__C": [0.1, 1, 10, 100],
```

```python
"svm__kernel": [
        "linear",
        "rbf",
        "poly"
    ],
```

```python
"svm__gamma": [
        "scale",
        "auto"
    ]
}
```

```python
grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)
```

```python
print("\nTraining GridSearchCV...")
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("\n====================================")
print("BEST PARAMETERS")
print("====================================")
```

```python
print(grid_search.best_params_)
```

```python
print("\nBest Cross Validation Score:")
print(grid_search.best_score_)
```

```python
best_model = grid_search.best_estimator_
```

```python
y_pred = best_model.predict(X_test)
```

```python
print("Accuracy :",
      accuracy_score(y_test, y_pred))
```

```python
print("Precision:",
      precision_score(y_test, y_pred, zero_division=0))
```

```python
print("Recall   :",
      recall_score(y_test, y_pred, zero_division=0))
```

```python
print("F1 Score :",
      f1_score(y_test, y_pred, zero_division=0))
```

```python
print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))
```

```python
print("\nResult:")
print("SVM hyperparameters were successfully tuned")
print("using GridSearchCV.")
```
