```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.cluster import DBSCAN
from sklearn.decomposition import PCA
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/Joerakesh/machine_learning_lab/refs/heads/main/datasets/heart_disease_dataset.csv")
```

```python
print("Dataset Shape:", df.shape)
print("\nFirst 5 records:")
print(df.head())
```

```python
X = df.drop("Heart Disease", axis=1)
```

```python
numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns
```

```python
categorical_features = X.select_dtypes(
    include=["object"]
).columns
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)
```

```python
X_processed = preprocessor.fit_transform(X)
```

```python
X_processed = X_processed
```

```python
dbscan = DBSCAN(
    eps=1.5,
    min_samples=5
)
```

```python
labels = dbscan.fit_predict(X_processed)
```

```python
df["Cluster"] = labels
```

```python
print("\nCluster Labels:")
print(df["Cluster"].value_counts().sort_index())
```

```python
noise_points = df[df["Cluster"] == -1]
```

```python
print("\nNumber of Noise Points:")
print(len(noise_points))
```

```python
print("\nNoise Points:")
print(noise_points.head(10))
```

```python
number_of_clusters = len(
    set(labels)
) - (1 if -1 in labels else 0)
```

```python
print("\nNumber of Clusters:")
print(number_of_clusters)
```

```python
pca = PCA(n_components=2)
```

```python
X_pca = pca.fit_transform(X_processed)
```

```python
plt.figure(figsize=(9, 6))
```

```python
plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=labels,
    cmap="viridis",
    s=40
)
```

```python
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("DBSCAN Clustering")
```

```python
plt.colorbar(label="Cluster")
```

```python
plt.show()
```
